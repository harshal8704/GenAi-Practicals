```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Check for GPU availability to speed up generation
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load the tokenizer and model
model_ckpt = "facebook/bart-large-cnn"
print(f"Downloading tokenizer and model for {model_ckpt}...")

tokenizer = AutoTokenizer.from_pretrained(model_ckpt)
model = AutoModelForSeq2SeqLM.from_pretrained(model_ckpt).to(device)
```

```python
# Sample lengthy news article
news_article = """
In a groundbreaking announcement on Tuesday, scientists at the Global Space Agency (GSA) revealed the discovery of an exoplanet
orbiting a red dwarf star just 40 light-years away from Earth. The planet, officially designated as Kepler-452c, resides in the
habitable zone of its star system, meaning temperatures could allow liquid water to pool on its surface.

Researchers utilized the James Webb Space Telescope to analyze the planet's atmospheric composition. Early data suggests the presence
of water vapor and methane, biosignatures that could indicate biological activity. Dr. Elena Rostova, the lead astrophysicist on the
project, stated that while this is a monumental leap in the search for extraterrestrial life, further observations are required over
the next decade to confirm the findings. The mission has received emergency funding from international space coalitions to fast-track
the deployment of deep-space monitoring satellites.
"""
print(f"Original Article Length: {len(news_article.split())} words")
```

```python
# 1. Tokenize the text (truncate if it exceeds the model's max limit of 1024 tokens)
inputs = tokenizer(news_article, max_length=1024, return_tensors="pt", truncation=True).to(device)

# 2. Generate the summary using Beam Search
summary_ids = model.generate(
    inputs["input_ids"],
    max_length=60,         # Maximum length of the summary
    min_length=20,         # Minimum length of the summary
    length_penalty=2.0,    # Encourages the model to generate longer sentences
    num_beams=4,           # Number of branches explored to find the best sequence
    early_stopping=True    # Stops generation when all beam hypotheses end
)

# 3. Decode the output tokens back into readable human text
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("\n--- GENERATED SUMMARY ---")
print(summary)
```
